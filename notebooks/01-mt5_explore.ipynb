```python
import pandas as pd
from dotenv import load_dotenv
import os


load_dotenv()

print(f"Account Info:, {os.getenv('ACCOUNT_BROKER')}")
```

```python
import MetaTrader5 as mt5

if not mt5.initialize(
    login=int(os.getenv("ACCOUNT_LOGIN_ID")),
    password=os.getenv("ACCOUNT_PASSWORD"),
    server=os.getenv("ACCOUNT_SERVER")
):
    print("initialize() failed, error code =", mt5.last_error())
    mt5.shutdown()
print(mt5.terminal_info())
print(mt5.version())
```

```python
for symbol in mt5.symbols_get():
    if symbol.visible:
        print(symbol.name, symbol.description, symbol.path)
```

```python
from datetime import datetime

datefrom = datetime(2019, 1, 1)

rates = mt5.copy_rates_from("EURUSD", mt5.TIMEFRAME_H4, datefrom, 500)
print(rates)
```

```python
current_date = datetime.now()
date_from = datetime(2019, 1, 1)

rates = mt5.copy_rates_range("EURUSD", mt5.TIMEFRAME_H4, date_from, current_date)

print(rates)
```
